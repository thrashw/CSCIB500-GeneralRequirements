To scrape the web page, you can use the following code. The get function gets a web page from a URL as a Response object, which has a content property containing the page’s content. You can then use Beautiful Soup to process the text and strip out the HTML elements.

```python
import requests
from bs4 import BeautifulSoup
response = requests.get('https://www.python.org')
soup = BeautifulSoup(response.content, 'html5lib')
text = soup.get_text(strip=True)
```
